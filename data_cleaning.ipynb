```python
import pandas as pd

df = pd.read_csv("mahasiswa_olah.csv")

print(df)
```

Lihat data yang kosong:

```python
print(df.isnull().sum())
```

Menghapus data kosong:

```python
df_bersih = df.dropna()

print(df_bersih)
```

Mengisi data kosong:

```python
df["Umur"] = df["Umur"].fillna(df["Umur"].mean())
df["Nilai"] = df["Nilai"].fillna(df["Nilai"].mean())

print(df)
```
