Grafik Batang

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('mahasiswa.csv')

plt.bar(df['Nama'], df['Nilai'])
plt.title('Nilai Mahasiswa')
plt.xlabel('Nama')
plt.ylabel('Nilai')
plt.show()
```

Grafik Garis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('mahasiswa.csv')

plt.plot(df['Nama'], df['Nilai'])
plt.title('Grafik Nilai Mahasiswa')
plt.show()
```

Histogram

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('mahasiswa.csv')

plt.hist(df['Nilai'])
plt.title('Distribusi Nilai')
plt.show()
```
