# Latihan Standar Deviasi

Dataset:
[4, 29, 98, 7, 96, 6]

Tujuan:
Menghitung rata-rata dan standar deviasi dari suatu dataset.

```python
data = [4, 29, 98, 7, 96, 6]

mean = sum(data) / len(data)

print("Mean =", mean)
```

```python
from statistics import stdev

data = [4, 29, 98, 7, 96, 6]

print("Standar Deviasi =", stdev(data))
```

```python
import matplotlib.pyplot as plt

data = [4, 29, 98, 7, 96, 6]

plt.hist(data)
plt.title("Distribusi Data")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```
