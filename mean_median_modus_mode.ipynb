```python
data = [3, 9, 1, 1, 2, 3, 9, 7, 3]

mean = sum(data) / len(data)

print("Mean:", mean)
```

```python
data = [3, 9, 1, 1, 2, 3, 9, 7, 3]

data.sort()

n = len(data)
median = data[n // 2]

print("Median =", median)
```

```python
from statistics import mode

data = [3, 9, 1, 1, 2, 3, 9, 7, 3]

print("Modus =", mode(data))
```
